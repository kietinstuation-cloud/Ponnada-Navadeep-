```python

# %% [markdown]
# # Project: Bridge Load Distribution Calculator (A2)
# **Course:** B.Tech Mathematics Mini Project
# **Concepts:** $AX = B$, Determinant, Matrix Inverse ($A^{-1}$), Static Equilibrium & Compatibility

# %% [markdown]
# ### Cell 1: Environment Setup and Library Imports

# %%
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
import pandas as pd
import ipywidgets as widgets
from ipywidgets import interact

# Plotting configuration
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 11
plt.rcParams['axes.grid'] = True
plt.grid(True, linestyle='--', alpha=0.6)

print("Cell 1 Complete: Environment configured successfully.")

# %% [markdown]
# ### Cell 2: Benchmark Input Data & Load Configurations

# %%
# Bridge Geometry (meters)
support_positions = np.array([0.0, 15.0, 30.0])  # Supports at x1, x2, x3

# Load Positions along the span (meters)
# Point 1: Pedestrian crowd at 5m
# Point 2: Initial Truck Position at 10m
# Point 3: Maintenance vehicle at 22m
load_positions = np.array([5.0, 10.0, 22.0])

# State 1 (Initial State): Loads in kiloNewtons (kN)
loads_state_1 = np.array([50.0, 120.0, 70.0])

# State 2 (Truck moves from 10m to 20m)
load_positions_state_2 = np.array([5.0, 20.0, 22.0])
loads_state_2 = np.array([50.0, 120.0, 70.0])

print(f"Bridge Span: {support_positions[-1]} m")
print(f"State 1 Loads: {loads_state_1} kN at positions {load_positions} m")
print(f"State 2 Loads: {loads_state_2} kN at positions {load_positions_state_2} m")
print("Cell 2 Complete: Inputs loaded.")

# %% [markdown]
# ### Cell 3: Mathematical Engine (Matrix Modeling, Determinant, Inverse)

# %%
def construct_system(support_x, load_x, loads):
    """
    Constructs the system A X = B using:
      Eq 1: Vertical force balance (sum(R) = sum(P))
      Eq 2: Moment balance about x = 0 (sum(x_i * R_i) = sum(p_j * P_j))
      Eq 3: Structural compatibility condition for continuous two-span beam
    """
    x1, x2, x3 = support_x

    # Coefficient matrix A (derived from boundary coordinates and elastic continuity)
    A = np.array([
        [1.0, 1.0, 1.0],
        [x1,  x2,  x3],
        [1.0, -4.0, 1.0]
    ], dtype=float)

    # Load Vector B
    b1 = np.sum(loads)
    b2 = np.sum(load_x * loads)

    # Normalized displacement influence term
    span = x3 - x1
    b3 = np.sum(loads * np.sin(np.pi * load_x / span))

    B = np.array([b1, b2, b3], dtype=float)
    return A, B

# 1. Symbolic verification of Invertibility via SymPy
A_sym = sp.Matrix([
    [1, 1, 1],
    [0, 15, 30],
    [1, -4, 1]
])
det_A_sym = A_sym.det()
A_inv_sym = A_sym.inv()

print("--- Symbolic Matrix Analysis ---")
print("Matrix A:")
sp.pprint(A_sym)
print(f"\nDeterminant det(A) = {det_A_sym}")
print("det(A) != 0 -> Matrix is non-singular; a UNIQUE solution exists.")
print("\nAnalytical Inverse A^(-1):")
sp.pprint(A_inv_sym)

# 2. Numerical Computation for State 1 and State 2
A_num, B1 = construct_system(support_positions, load_positions, loads_state_1)
_, B2 = construct_system(support_positions, load_positions_state_2, loads_state_2)

# Compute numerical inverse: X = A^(-1) * B
A_inv = np.linalg.inv(A_num)
R_state_1 = A_inv @ B1
R_state_2 = A_inv @ B2
delta_R = R_state_2 - R_state_1

print("\nCell 3 Complete: Mathematical engine executed.")

# %% [markdown]
# ### Cell 4: Tabular Output Display & Equilibrium Check

# %%
results_df = pd.DataFrame({
    'Support': ['R1 (Left @ 0m)', 'R2 (Center @ 15m)', 'R3 (Right @ 30m)'],
    'State 1 Reaction (kN)': R_state_1,
    'State 2 Reaction (kN)': R_state_2,
    'Delta Change (kN)': delta_R,
    'Percentage Change (%)': (delta_R / R_state_1) * 100
})

print("=" * 72)
print("       BRIDGE LOAD DISTRIBUTION CALCULATOR - REACTION REPORT")
print("=" * 72)
print(results_df.round(2).to_string(index=False))
print("-" * 72)

# Static Equilibrium Verification: Sum of Reactions must equal Sum of Applied Loads
total_load_1, total_react_1 = np.sum(loads_state_1), np.sum(R_state_1)
total_load_2, total_react_2 = np.sum(loads_state_2), np.sum(R_state_2)

print(f"Equilibrium Verification (State 1): sum(Fy) = {total_react_1:.2f} kN | Applied Load = {total_load_1:.2f} kN (Error: {abs(total_react_1 - total_load_1):.1e})")
print(f"Equilibrium Verification (State 2): sum(Fy) = {total_react_2:.2f} kN | Applied Load = {total_load_2:.2f} kN (Error: {abs(total_react_2 - total_load_2):.1e})")
print("=" * 72)
print("Cell 4 Complete: Results tabulated and verified.")

# %% [markdown]
# ### Cell 5: Visual Engineering Analytics (Static Comparison)

# %%
fig, axes = plt.subplots(2, 1, figsize=(13, 10))

# Subplot 1: Bridge Deck Schematic
ax1 = axes[0]
ax1.plot([0, 30], [0, 0], color='dimgray', linewidth=8, label='Bridge Deck (30m)')

for pos, name in zip(support_positions, ['R1', 'R2', 'R3']):
    ax1.plot(pos, -0.05, marker='^', markersize=16, color='black')
    ax1.text(pos, -0.15, name, ha='center', fontweight='bold')

ax1.quiver(load_positions, np.zeros_like(load_positions), 0, -loads_state_1,
           color='teal', angles='xy', scale_units='xy', scale=400, width=0.008, label='State 1 Applied Loads')
ax1.quiver(load_positions_state_2, np.zeros_like(load_positions_state_2), 0, -loads_state_2,
           color='coral', angles='xy', scale_units='xy', scale=400, width=0.005, label='State 2 Applied Loads (Shifted Truck)')

ax1.set_ylim(-0.35, 0.2)
ax1.set_xlim(-2, 32)
ax1.set_title("Bridge Configuration: Physical Load Migration along the Span", fontsize=14, fontweight='bold')
ax1.set_xlabel("Bridge Span Position (meters)")
ax1.get_yaxis().set_visible(False)
ax1.legend(loc='upper right')

# Subplot 2: Comparative Bar Chart
ax2 = axes[1]
x_indices = np.arange(len(support_positions))
width = 0.35

bars1 = ax2.bar(x_indices - width/2, R_state_1, width, label='State 1 Reactions', color='steelblue')
bars2 = ax2.bar(x_indices + width/2, R_state_2, width, label='State 2 Reactions', color='darkorange')

for bar in bars1:
    yval = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2, yval + 1.5, f"{yval:.1f} kN", ha='center', fontsize=9)
for bar in bars2:
    yval = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2, yval + 1.5, f"{yval:.1f} kN", ha='center', fontsize=9)

ax2.set_xticks(x_indices)
ax2.set_xticklabels(['Support 1 (x=0m)', 'Support 2 (x=15m)', 'Support 3 (x=30m)'])
ax2.set_ylabel("Reaction Force (kN)")
ax2.set_title("Reaction Redistribution: How Moving One Vehicle Changes All Three Supports", fontsize=14, fontweight='bold')
ax2.legend()

plt.tight_layout()
plt.show()
print("Cell 5 Complete: Static plots generated.")

# %% [markdown]
# ### Cell 6: Interactive Truck Movement Simulation (`ipywidgets`)

# %%
# Precompute continuous reaction histories across the span
x_span = np.linspace(0.0, 30.0, 150)
R1_history, R2_history, R3_history = [], [], []

P1_mag, P1_pos = 50.0, 5.0
P3_mag, P3_pos = 70.0, 22.0
truck_load_default = 120.0

for x_pos in x_span:
    current_loads = np.array([P1_mag, truck_load_default, P3_mag])
    current_positions = np.array([P1_pos, x_pos, P3_pos])
    _, B_eval = construct_system(support_positions, current_positions, current_loads)
    R_eval = A_inv @ B_eval
    R1_history.append(R_eval[0])
    R2_history.append(R_eval[1])
    R3_history.append(R_eval[2])

R1_history = np.array(R1_history)
R2_history = np.array(R2_history)
R3_history = np.array(R3_history)


def update_bridge_simulation(truck_x=10.0, truck_weight=120.0):
    curr_loads = np.array([P1_mag, truck_weight, P3_mag])
    curr_positions = np.array([P1_pos, truck_x, P3_pos])

    _, B_curr = construct_system(support_positions, curr_positions, curr_loads)
    R_curr = A_inv @ B_curr

    fig, (ax_phys, ax_lines) = plt.subplots(2, 1, figsize=(13, 8), gridspec_kw={'height_ratios': [1, 1.4]})

    # Top: Schematic
    ax_phys.plot([0, 30], [0, 0], color='#4A5568', linewidth=9, zorder=2)

    for pos, label in zip(support_positions, ['R1 (0m)', 'R2 (15m)', 'R3 (30m)']):
        ax_phys.plot(pos, -0.05, marker='^', markersize=14, color='#1A202C', zorder=3)
        ax_phys.text(pos, -0.16, label, ha='center', fontweight='bold', fontsize=10)

    ax_phys.quiver([P1_pos, P3_pos], [0, 0], [0, 0], [-P1_mag, -P3_mag],
                   color='#2B6CB0', angles='xy', scale_units='xy', scale=350,
                   width=0.008, label='Stationary Loads (P1, P3)', zorder=4)

    ax_phys.quiver([truck_x], [0], [0], [-truck_weight],
                   color='#E53E3E', angles='xy', scale_units='xy', scale=350,
                   width=0.012, label=f'Moving Truck ({truck_weight} kN)', zorder=5)

    ax_phys.text(0.0, 0.08, f"R1 = {R_curr[0]:.1f} kN", color='#2C5282', fontweight='bold', ha='center')
    ax_phys.text(15.0, 0.08, f"R2 = {R_curr[1]:.1f} kN", color='#276749', fontweight='bold', ha='center')
    ax_phys.text(30.0, 0.08, f"R3 = {R_curr[2]:.1f} kN", color='#744210', fontweight='bold', ha='center')

    ax_phys.set_xlim(-2, 32)
    ax_phys.set_ylim(-0.3, 0.22)
    ax_phys.set_title(f"Dynamic Position: Truck at x = {truck_x:.1f} m (Total Load = {np.sum(curr_loads):.1f} kN)",
                      fontsize=13, fontweight='bold')
    ax_phys.get_yaxis().set_visible(False)
    ax_phys.legend(loc='upper right', framealpha=0.9)

    # Bottom: Reaction profiles across span
    ax_lines.plot(x_span, R1_history, label='Support 1 (R1)', color='#3182CE', linewidth=2.2)
    ax_lines.plot(x_span, R2_history, label='Support 2 (R2)', color='#38A169', linewidth=2.2)
    ax_lines.plot(x_span, R3_history, label='Support 3 (R3)', color='#D69E2E', linewidth=2.2)

    ax_lines.scatter([truck_x], [R_curr[0]], color='#3182CE', s=80, zorder=5)
    ax_lines.scatter([truck_x], [R_curr[1]], color='#38A169', s=80, zorder=5)
    ax_lines.scatter([truck_x], [R_curr[2]], color='#D69E2E', s=80, zorder=5)
    ax_lines.axvline(truck_x, color='#E53E3E', linestyle=':', alpha=0.8, label='Truck Position')

    ax_lines.set_xlabel("Truck Position along Bridge Span (m)", fontsize=11)
    ax_lines.set_ylabel("Reaction Force (kN)", fontsize=11)
    ax_lines.set_title("Support Reactions vs. Moving Truck Position", fontsize=12, fontweight='bold')
    ax_lines.set_xlim(0, 30)
    ax_lines.legend(loc='center right', framealpha=0.9)

    plt.tight_layout()
    plt.show()

truck_slider = widgets.FloatSlider(
    value=10.0, min=0.0, max=30.0, step=0.2,
    description='Truck x (m):', continuous_update=True, readout_format='.1f',
    layout=widgets.Layout(width='60%')
)

weight_slider = widgets.FloatSlider(
    value=120.0, min=40.0, max=250.0, step=5.0,
    description='Load (kN):', continuous_update=True, readout_format='.0f',
    layout=widgets.Layout(width='60%')
)

interact(update_bridge_simulation, truck_x=truck_slider, truck_weight=weight_slider);
print("Cell 6 Complete: Interactive widget initialized.")
```
